Multimodal Model Exploration

This notebook demonstrates different modalities and model providers using `vikas_app.py`. 
The notebook is built under `Assignments/Assignment3_MultiModality_Model` and uses free provider keys from `.env`.

```python
import os
import sys
from pathlib import Path
from IPython.display import display, Image, Audio, HTML

project_dir = Path.cwd() / "Assignments" / "Assignment3_MultiModality_Model"
sys.path.insert(0, str(project_dir))

from vikas_app import MultimodalApp

app = MultimodalApp()
print("Loaded multimodal app successfully.")
```

## Text → Text

Using OpenRouter or Hugging Face for text generation.

```python
prompt = "Explain how multimodal AI can improve learning experiences for students."
app.text_model.provider = "openrouter"
output_text = app.text_to_text(prompt)
print("Prompt:", prompt)
print("
Generated Text:")
print(output_text)
```

## Text → Image

Generate an image from a text prompt using Stable Diffusion.

```python
image_prompt = "A futuristic classroom where students and robots collaborate in a bright science lab."
generated_image = app.text_to_image(image_prompt)
if generated_image is not None:
    generated_image.save("assignment3_text_to_image.png")
    display(generated_image)
else:
    print("Image generation failed.")
```

## Image → Text

Caption the generated image or any uploaded image.

```python
image_path = "assignment3_text_to_image.png"
if os.path.exists(image_path):
    caption = app.image_to_text(image_path)
    print("Image caption:")
    print(caption)
else:
    print("No generated image found. Run the Text → Image cell first.")
```

## Image → Image

Apply a creative transformation to an existing image using image-to-image diffusion.

```python
base_image_path = "assignment3_text_to_image.png"
if os.path.exists(base_image_path):
    transform_prompt = "Turn this scene into a bright storybook illustration."
    transformed = app.image_to_image(base_image_path, transform_prompt)
    if transformed is not None:
        transformed.save("assignment3_image_to_image.png")
        display(transformed)
    else:
        print("Image transformation failed.")
else:
    print("Base image not found. Run Text → Image first.")
```

## Text → Audio

Generate speech audio from a text prompt.

```python
audio_text = "Welcome to Assignment 3, where we explore multimodal AI across text, image, audio, and video."
audio_output = app.text_to_audio(audio_text, output_path="assignment3_text_to_audio.wav")
print("Audio output path:", audio_output)
if os.path.exists(audio_output):
    display(Audio(audio_output))
else:
    print("Text-to-audio generation failed.")
```

## Audio → Text

Transcribe audio to text using a speech recognition model.

```python
audio_file = "assignment3_text_to_audio.wav"
if os.path.exists(audio_file):
    transcript = app.audio_to_text(audio_file)
    print("Transcript:")
    print(transcript)
else:
    print("Audio file not found. Run Text → Audio first.")
```

## Audio → Audio

Convert an audio file through transcription and re-synthesis.

```python
audio_input = "assignment3_text_to_audio.wav"
if os.path.exists(audio_input):
    converted_audio = app.audio_to_audio(audio_input, output_path="assignment3_audio_to_audio.wav")
    print("Converted audio path:", converted_audio)
    if os.path.exists(converted_audio):
        display(Audio(converted_audio))
    else:
        print("Audio→Audio conversion failed.")
else:
    print("Source audio not found. Run Text → Audio first.")
```

## Text → Video

Generate a short video from a text prompt by composing multiple generated frames.

```python
video_prompt = "A cinematic montage of AI and education, with bright colors and inspiring tone."
video_path = app.text_to_video(video_prompt, output_path="assignment3_text_to_video.mp4")
print("Video path:", video_path)
if os.path.exists(video_path):
    display(HTML(f"<video width=640 controls><source src='{video_path}' type='video/mp4'></video>"))
else:
    print("Text-to-video generation failed.")
```

## Video → Text

Extract captions from video frames to produce a text summary.

```python
video_file = "assignment3_text_to_video.mp4"
if os.path.exists(video_file):
    captions = app.video_to_text(video_file)
    print("Video captions:")
    for idx, caption in enumerate(captions, start=1):
        print(f"Frame {idx}: {caption}")
else:
    print("Video file not found. Run Text → Video first.")
```

## Model Summary

- **Text → Text**: OpenRouter GPT-4o-mini or Hugging Face GPT-Neo
- **Text → Image**: Stable Diffusion `runwayml/stable-diffusion-v1-5`
- **Image → Text**: `nlpconnect/vit-gpt2-image-captioning`
- **Image → Image**: Stable Diffusion img2img
- **Text → Audio**: `espnet/kan-bayashi_ljspeech_vits`
- **Audio → Text**: `facebook/wav2vec2-base-960h`
- **Text → Video**: Frame-based video generation using Stable Diffusion
- **Video → Text**: Video frame captioning and summarization
